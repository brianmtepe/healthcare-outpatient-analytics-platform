```python
# Outpatient Trend Analysis

## Phase 4

Data Cleaning Pipeline

Purpose

Transform raw synthetic healthcare datasets into clean,
validated datasets suitable for SQL database loading.
```

```python
import pandas as pd
import numpy as np
import os

print("Libraries loaded successfully.")
```

```python
RAW_PATH = "../Data/raw"

CLEAN_PATH = "../Data/cleaned"

os.makedirs(CLEAN_PATH, exist_ok=True)

print("Project directories verified.")
```

```python
# load datasets

df_departments = pd.read_csv(f"{RAW_PATH}/departments.csv")
df_diagnoses = pd.read_csv(f"{RAW_PATH}/diagnoses.csv")
df_providers =pd.read_csv(f"{RAW_PATH}/providers.csv")
df_patients = pd.read_csv(f"{RAW_PATH}/patients.csv")
df_visits =pd.read_csv(f"{RAW_PATH}/visits.csv")

print("Datasets successfully loaded")
```

```python
# Data Profiling

datasets = {
    "Departments": df_departments, 
    "Diagnoses": df_diagnoses,
    "Providers": df_providers,
    "Patients": df_patients,
    "Visits": df_visits
}

for name, df in datasets.items():
    print ("="*60)
    print (name)
    print ("="*60)

    print (df.info())
    print (df.head())
    print (df.shape)
```

```python
# Missing Values Assessment 

for name, df in datasets.items():

    print ("="*50)
    print (name)
    print (df.isnull().sum())
```

```python
# Duplicate Detection

for name, df in datasets.items():

    print("="*50)
    print (name)
    print ("Duplicate rows:", df.duplicated().sum())
```

```python
# Data type Verification

print (df_patients.dtypes)
print (df_visits.dtypes)
```

```python
# Convert Date Columns

df_patients["registration_date"] = pd.to_datetime(df_patients["registration_date"])
df_patients["date_of_birth"] = pd.to_datetime(df_patients["date_of_birth"])
df_visits["visit_date"] =pd.to_datetime(df_visits["visit_date"])

print ("Date columns converted")
```

```python
# Relationship Validation

print(
    df_visits["patient_id"].isin( df_patients["patient_id"]).all()
)
print(
    df_visits["provider_id"].isin(df_providers["provider_id"]).all()
)
print(
    df_visits["department_id"].isin(df_departments["department_id"]).all()
)
print(
    df_visits["diagnosis_id"].isin(df_diagnoses["diagnosis_id"]).all()
)
```

```python
# Export Clean Data

df_departments.to_csv(f"{CLEAN_PATH}/departments_clean.csv", index=False)
df_diagnoses.to_csv(f"{CLEAN_PATH}/diagnoses_clean.csv", index=False)
df_providers.to_csv(f"{CLEAN_PATH}/providers_clean.csv", index=False)
df_patients.to_csv(f"{CLEAN_PATH}/patients_clean.csv", index=False)
df_visits.to_csv(f"{CLEAN_PATH}/visits_clean.csv", index=False)

print ("Cleaned datasets exported successfully")
```
