```python
import pandas as pd
import numpy as np
import os

print("Libraries imported successfully")
```

```python
# Load Processed Files

PROCESSED_PATH = "../Data/processed"
df_departments = pd.read_csv(f"{PROCESSED_PATH}/departments_processed.csv")
df_diagnoses = pd.read_csv(f"{PROCESSED_PATH}/diagnoses_processed.csv")
df_providers = pd.read_csv(f"{PROCESSED_PATH}/providers_processed.csv")
df_patients = pd.read_csv(f"{PROCESSED_PATH}/patients_processed.csv")
df_visits = pd.read_csv(f"{PROCESSED_PATH}/visits_processed.csv")

print("Processed datasets loaded")
                        
```

```python
# Dataset Dimensions
datasets={
    "Departments": df_departments,
    "Diagnoses": df_diagnoses,
    "Providers": df_providers,
    "Patients": df_patients,
    "Visits": df_visits
}
for name ,df in datasets.items():
    print("="*60)
    print(name)
    print("="*60)
    print(f"Rows: {df.shape[0]:,}")
    print(f"Columns: {df.shape[1]}")
```

```python
# Null Values
for name, df in datasets.items():

    print("=" * 60)

    print(name)

    print("=" * 60)

    print(df.isnull().sum())
```

```python
# Duplicates
for name, df in datasets.items():

    print("=" * 60)

    print(name)

    print("=" * 60)

    print("Duplicates:", df.duplicated().sum())
```

```python
# Primary Key Validation
print(df_patients["patient_id"].is_unique)

print(df_providers["provider_id"].is_unique)

print(df_departments["department_id"].is_unique)

print(df_diagnoses["diagnosis_id"].is_unique)

print(df_visits["visit_id"].is_unique)
```

```python
# Foreign Key Validation
print(
    df_visits["patient_id"].isin(
        df_patients["patient_id"]
    ).all()
)

print(
    df_visits["provider_id"].isin(
        df_providers["provider_id"]
    ).all()
)

print(
    df_visits["department_id"].isin(
        df_departments["department_id"]
    ).all()
)

print(
    df_visits["diagnosis_id"].isin(
        df_diagnoses["diagnosis_id"]
    ).all()
)
```

```python
# Business Rule Validation

print("Negative Waiting Time:",(df_visits["waiting_time_minutes"]<0).sum())
print("Negative COnsultation Time:",(df_visits["consultation_minutes"]<0).sum())
print("Negative Bill Amount:",(df_visits["bill_amount"]<0).sum())
```

```python
# Age Validation

print(df_patients["age"].describe())
```

```python
# Visit Date Validation

print("Earliest Visits:",df_visits["visit_date"].min())
print("Latest Visit:",df_visits["visit_date"].max())
```

```python
VALIDATED_PATH = "../Data/validated"

os.makedirs(VALIDATED_PATH, exist_ok=True)

df_departments.to_csv(
    f"{VALIDATED_PATH}/departments.csv",
    index=False
)

df_diagnoses.to_csv(
    f"{VALIDATED_PATH}/diagnoses.csv",
    index=False
)

df_providers.to_csv(
    f"{VALIDATED_PATH}/providers.csv",
    index=False
)

df_patients.to_csv(
    f"{VALIDATED_PATH}/patients.csv",
    index=False
)

df_visits.to_csv(
    f"{VALIDATED_PATH}/visits.csv",
    index=False
)

print("Validated datasets exported successfully.")
```
